```python
data = pd.read_pickle('reconciled_data.pkl')
```

```python
data_FTS = pd.DataFrame(columns = ['index', 'W', 'Tin', 'Tout', 'Pin', 'Pout'])
```

```python
data_FTS.Tin = data.T_R_K
data_FTS.Pin = data.P_abs_Pa
data_FTS.W = data.m_cat_g
data_FTS['Ta'] = data.T_R_K
data_FTS['Fin_hydrogen'] = data.F_H2_e_mol_s
data_FTS['Fin_carbon monoxide'] = data.F_CO_e_mol_s
```

```python
data_FTS = data[[
    'F_H2_s_mol_s', 'F_CH4_s_mol_s', 'F_CO_s_mol_s', 'F_CO2_s_mol_s',
       'F_C2H4_s_mol_s', 'F_C2H6_s_mol_s', 'F_C3H8_s_mol_s',
    'F_H2O_s_mol_s',
    'F_C9', 'F_C10', 'F_C11', 'F_C12', 'F_C13', 'F_C14', 'F_C15', 'F_C16',
       'F_C17', 'F_C18', 'F_C19', 'F_C20', 'F_C21', 'F_C22', 'F_C23', 'F_C24',
       'F_C25', 'F_C26', 'F_C27', 'F_C28', 'F_C29', 'F_C30',
       'F_Olef_C9', 'F_Olef_C10', 'F_Olef_C11', 'F_Olef_C12', 'F_Olef_C13',
       'F_Olef_C14', 'F_Olef_C15',
    'T_R_K', 'P_abs_Pa', 'F_H2_e_mol_s', 'F_CO_e_mol_s', 'm_cat_g'
]]
```

```python
data_FTS = data_FTS.rename(columns = {
    'F_H2_s_mol_s'      : 'Fout_hydrogen',
    'F_CH4_s_mol_s'     : 'Fout_methane',
    'F_CO_s_mol_s'      : 'Fout_carbon monoxide',
    'F_CO2_s_mol_s'     : 'Fout_carbon dioxide',
    'F_C2H4_s_mol_s'    : 'Fout_ethylene',
    'F_C2H6_s_mol_s'    : 'Fout_ethane',
    'F_C3H8_s_mol_s'    : 'Fout_propane',
    'F_H2O_s_mol_s'     : 'Fout_water',
    'F_C9'              : 'Fout_nonane',
    'F_C10'             : 'Fout_decane',
    'F_C11'             : 'Fout_undecane',
    'F_C12'             : 'Fout_dodecane',
    'F_C13'             : 'Fout_tridecane',
    'F_C14'             : 'Fout_tetradecane',
    'F_C15'             : 'Fout_pentadecane',
    'F_C16'             : 'Fout_hexadecane',
    'F_C17'             : 'Fout_heptadecane',
    'F_C18'             : 'Fout_octadecane',
    'F_C19'             : 'Fout_nonadecane',
    'F_C20'             : 'Fout_eicosane',
    'F_C21'             : 'Fout_heneicosane',
    'F_C22'             : 'Fout_docosane',
    'F_C23'             : 'Fout_tricosane',
    'F_C24'             : 'Fout_tetracosane',
    'F_C25'             : 'Fout_pentacosane',
    'F_C26'             : 'Fout_hexacosane',
    'F_C27'             : 'Fout_heptacosane',
    'F_C28'             : 'Fout_octacosane',
    'F_C29'             : 'Fout_nonacosane',
    'F_C30'             : 'Fout_triacontane',
    'F_Olef_C9'         : 'Fout_1-nonene',
    'F_Olef_C10'        : 'Fout_1-decene',
    'F_Olef_C11'        : 'Fout_1-undecene',
    'F_Olef_C12'        : 'Fout_1-dodecene',
    'F_Olef_C13'        : 'Fout_1-tridecene',
    'F_Olef_C14'        : 'Fout_1-tetradecene',
    'F_Olef_C15'        : 'Fout_1-pentadecene',
    'T_R_K'             : 'Tin',
    'P_abs_Pa'          : 'Pin',
    'F_H2_e_mol_s'      : 'Fin_hydrogen',
    'F_CO_e_mol_s'      : 'Fin_carbon monoxide',
    'm_cat_g'           : 'W'
})
```

```python
data_FTS = data_FTS.astype('object')
```

```python
data_FTS.loc[0] = data_FTS.columns
data_FTS.loc[0] = data_FTS.loc[0].str.removeprefix('Fout_')
data_FTS.loc[0] = data_FTS.loc[0].str.removeprefix('Fin_')
data_FTS.loc[1] = 'mol/s'
```

```python
data_FTS.loc[0,'index'] = 'component names'
data_FTS.loc[1] = 'mol/s'
data_FTS.loc[1,'index'] = 'unit'
data_FTS.loc[1, 'Tin'] = 'K'
data_FTS.loc[1, 'Ta'] = 'K'
data_FTS.loc[1, 'Pin'] = 'Pa'
data_FTS.loc[1, 'W'] = 'g'
```

```python
data_FTS = data_FTS.sort_index()
```

```python
data_FTS.to_pickle('data_FTS.pkl')
```
