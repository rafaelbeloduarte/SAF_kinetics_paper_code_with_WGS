```python
MM = {}
for key in n_carbons.keys():
    if double_bond[key]:
        MM[key] = 12 * n_carbons[key] + 2 * n_carbons[key]
    else:
        MM[key] = 12 * n_carbons[key] + 2 * n_carbons[key] + 2
MM['Carbon dioxide'] = 44
MM['Water'] = 18
MM['Hydrogen'] = 2
MM['Carbon monoxide'] = 28
```

```python
cols_Fout_C1_C4 = [key for key, value in n_carbons.items() if value in range(1,5)]
cols_Fout_C5_C8 = [key for key, value in n_carbons.items() if value in range(5,9)]
cols_Fout_C9_C15 = [key for key, value in n_carbons.items() if value in range(9,16)]
cols_Fout_C16p = [key for key, value in n_carbons.items() if value in range(16,37)]
cols_Fout_C5p = [key for key, value in n_carbons.items() if value in range(5,37)]
```

```python
data = pd.read_excel('TOC_graphic_profiles.ods')
```

```python
cols_MolConc = [f'{comp} MolConc (mol/m3)' for comp in components]
```

```python
cols_MolConc.append('Length (m)')
```

```python
data[cols_MolConc].melt(id_vars = 'Length (m)')
```

```python
fig, ax = plt.subplots(figsize=(4.5,3.5))

sns.lineplot(data = data[cols_MolConc].melt(id_vars = 'Length (m)'),
            x = 'Length (m)', y = 'value', 
             hue = 'variable', style = 'variable', ax = ax, palette = 'bright'
            )
plt.grid('both')
plt.yscale('log')
plt.xlabel('Reactor length (m)', fontsize='large')
plt.ylabel(r'Molar fraction ($\mathrm{mol~m^{-3}}$)', fontsize='large')

# Get all handles and labels automatically generated
handles, labels = ax.get_legend_handles_labels()
labels = [item.replace(' MolConc (mol/m3)', '') for item in labels]

# Show only the first 10 items in the legend
ax.legend(handles[:12], labels[:12], bbox_to_anchor=(1.02, 1.02))

plt.tight_layout()
fig_name = 'TOC_graphic.svg'
plt.savefig(fig_name, 
            bbox_inches = 'tight')

plt.show()
```
